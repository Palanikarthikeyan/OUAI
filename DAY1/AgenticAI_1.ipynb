```python
## PromptTemplate + LLM
# -------------------------------

from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq

# llm object
llm_obj = ChatGroq(model="openai/gpt-oss-20b",api_key=myapi_key)

# prompt object
pobj = PromptTemplate(input_variables = ['topic','level'],template = """Explain about {topic} for a {level} student.
Give me a simple example""")

# create LangChainExpression Language
chain = pobj | llm_obj

response = chain.invoke({'topic':'Perl','level':'beginner'})
print(response.content)
```

```python
### Prompt -> LLM -->Answer
# -------------------------------// GenAI
#
# Agentic AI
# ----------------
# LLM + tool
# ==== -------
#
# langchain - tool
# -----------------
#   |->python decorator - function - adding more features to an exisiting code
#
#  @<dectorfunction>
#  def functionName(): <=== act as a tool
#      ..... code logic
# -------------------------

# agent = LLM + tool + Prompt
#          ---  =====   ------
#  |
# Invoke agent
# ----------------------------------------------

# Tool - function - llm/agent can call to perform an action
#
# User ->|AI Agent(Application)
# ------ |-----------------------
#        | Decides - I need weather information
#        |      |____weather tool
#                    -----------------
#                    |__ weather tool API --->tool -->result --->LLM(observation) -->Final result
#
#
# Requirement
#  |->current weather ---------> weather API
#  |->Search internet ----------------->  Web search
#  |->Calculate 2500 * 12 ---------------> calculator
#                                           ......
# .......................... ===========================
```

```python
from langchain_core.tools import tool

@tool
def calc(a: int,b: int) ->int:
  '''Add two numbers'''
  return a+b
```

```python
from langchain.agents import create_agent
# create_agent(model=<llmobject>,tools = [tool_name1,tool_name2,...]) -> agent_object
#                    ==========
## agent_object.invoke({'messages':[{'role':'User','content':'User prompt'}]})
#
```

```python
## LLM+Tool
# ------------------
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
# llm object
llm_obj = ChatGroq(model="openai/gpt-oss-20b",api_key=myapi_key)

# Create a tool
# -----------------
@tool
def mycalc(expression: str) -> str:
  '''simple calculation'''
  return str(eval(expression))


# Create Agent = LLM + tool
agent_obj = create_agent(model = llm_obj,tools = [mycalc])

agent_obj.invoke({'messages':[{'role':'user','content':"Calculate 10 + 20 and explain the result"}]})
```

```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent

# llm object
llm_obj = ChatGroq(model="openai/gpt-oss-20b",api_key=myapi_key)

# Create a tool
# -----------------
@tool
def mycalc(expression: str) -> str:
  '''simple calculation'''
  return str(eval(expression))


# Create Agent = LLM + tool
agent_obj = create_agent(model = llm_obj,tools = [mycalc])

result = agent_obj.invoke({'messages':[{'role':'user','content':"Calculate 10 + 20 and explain the result"}]})
print(result)
```

```python
print("User Query:",result['messages'][0].content)
```

```python
print("AI Response:",result['messages'][1].content)
```

```python
print("Tool Result:",result['messages'][2].content)
```
