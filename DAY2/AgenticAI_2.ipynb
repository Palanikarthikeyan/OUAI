```python
# To implement Chat Conversation
#
# 1. ChatMessageHistory   - Store conversation messages between user and AI
# 2. BaseChatMessageHistory  - interface - define - how the chat history should stored and managed
# 3. RunnableWithMessageHistory - wrapper class

# User: My name is Karthik
# AI: Hello Karthik Nice to Meet you

# User: What is my name?
# AI: I don't know

# User: My name is Karthik
#       |
#   Chat History
#       |
#   HumanMessage("My name is Karthik")
#
# # User: What is my name?
#  History + NewQuestion
#      |
#   LLM
#
#


from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
stores = {}

def get_session_history(session_id: str) -> BaseChatMessageHistory:
  if session_id not in stores:
    stores[session_id] = ChatMessageHistory()
  return stores[session_id]
```

```python
## multi-tools
## ------------
##   tool-1: Calc
##   tool-2: get_weather
##   tool-3: db



# User Input  ->RunnableLambda -->Prepare input -->create_agent() ->LLM + Tools
#                                                  =============================//agent_chain
#                                 Prepare_input|agent_obj ->agent_chain


from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_core.runnables import RunnableLambda


# llm object
llm_obj = ChatGroq(model="openai/gpt-oss-20b",api_key=myapi_key)

# Create a tool
# -----------------
@tool
def mycalc(expression: str) -> str:
  '''simple calculation'''
  return str(eval(expression))

@tool
def get_weather(city: str) ->str:
  '''Get the current weather for a city'''
  weather_data = {
      "Bengaluru": "24C, Cloudy",
      "Mumbai":"29C humid",
      "London":"16C Cloudy"
  }
  return weather_data.get(city,f"No weather data availalbe for {city}")

@tool
def search_db(query: str) -> str:
  '''search our company customer data'''
  customers = {"leo":{"plan":"enterprise","usage":"89%"},
               "anu":{"plan":"prod","usage":"61%"}
               }
  query_lower = query.lower()
  for name,data in customers.items():
    if name in query_lower:
      return (f'''Customer:
      {name.title()},
      f"Plan: {data['plan']},
      f"Usage:{data['usage']}''')
  return "Customer Not Found"


# Create Agent = LLM + tool
my_tools = [mycalc,get_weather,search_db]

agent_obj = create_agent(model = llm_obj,tools = my_tools,system_prompt="""
You are a helpful agent.
You have acccess to multiple tools
""")

## Runnable lambda

def prepare_input(user_input):
  '''Convert normal input to message expected by the agent'''
  return {
      "messages":[{
          'role':"user",
          'content':user_input.strip()
      }]
  }


prepare = RunnableLambda(prepare_input)

# Combine RunnableLambda + Agent
agent_chain = prepare |agent_obj

#
#result = agent_obj.invoke({'messages':[{'role':'user','content':"Calculate 10 + 20 and explain the result"}]})
#print(result)
#

# Invoke
#-------------
response = agent_chain.invoke("what is 10+20 and what's the weather in Begaluru?")
print(response)
```

```python
## Retrieval Augmented Generation (RAG)
# --------------------------------------------
#  Technique that allows an LLM to retrieve relevant information from our own data
#  -------------------------------------------------------------------------------
#  RAG = Search our data + Give the results to the LLM + Generate an Answer
# -----
## Without RAG
#
#  User -> LLM(llm_obj) ->Answer

# User: HR Policy PDF details / last week sales report / get last 2 weeks error logs from /var/log directory
#      ====================================================================================================
#                   |--> LLM may not know our information(HR Policy,sales records,/var/log/..)
#
#
#  Our Documents (PDF/Excel/Folder/logs/DB/Web/JSON etc.,)
#   |
#  Split into chunks
#   |
#  Embeddings  - each chunk is converted to number - embedding (semantic meaning)
#   |
#  Vectors - numerical array
#   |
# Vector DataBase
#  |
#  Retriever <--- User Question
#  ||
#  Relevant documents
#  ||
#  LLM
#  || <-- Apply Prompt
#  Answer
#######################


# 1.  Data Loader  ->[metadata,page_content]
#     ============
#           |-> meta data                +    page_content
#              info about loaded data           |->Actual data
#
# 2. split data into multiplechunks - chunkSize,chunkOverlap
#
# 3. Create embedding model - object
# |
# 4. Store to VectorDatabase (FAISS,Chroma,pinechore,Oracle23ai)
# |
# 5. similarity search
# -------------------------------------// data indexing
#
# 6. Create retriever object
# 7. create llm object
# 8. create prompt object
# 9. Retrieval QA Chain = combine retriever object + llm object + prompt object -> RAGchain
# |
# 10. RAGChain.invoke
# -------------------------------------------------
```
